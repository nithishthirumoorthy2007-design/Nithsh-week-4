```python
import pandas as pd
df = pd.read_csv("SHOP_2015-05-21_2025-03-16.csv")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df["Date"] = pd.to_datetime(df["date"])
```

```python
df = df.sort_values("Date")
```

```python
df  = df.set_index("Date")
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=df,
    x=df.index,
    y="close"
)

plt.title("Shopify Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("closing price")

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=df,
    x=df.index,
    y="open"
)

plt.title("Shopify Opening Price Over Time")
plt.xlabel("Date")
plt.ylabel("Opening Price")

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(
    data=df,
    x=df.index,
    y="open",
    label="Open Price"
)

sns.lineplot(
    data=df,
    x=df.index,
    y="close",
    label="Close Price"
)

plt.title("Shopify Open and Close Price Over Time")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()
plt.show()
```

```python
#visualize OHLC price

plt.figure(figsize=(14,6))

sns.lineplot(data=df, x=df.index, y="open", label="open")
sns.lineplot(data=df, x=df.index, y="high", label="high")
sns.lineplot(data=df, x=df.index, y="low", label="low")
sns.lineplot(data=df, x=df.index, y="close", label="close")

plt.title("Shopify Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("price")

plt.show()
```

```python
plt.figure(figsize=(14,5))

sns.lineplot(
    data=df,
    x="Date",
    y="volume"
)

plt.title("Shopify Trading Over Time")
plt.xlabel("Date")
plt.ylabel("Trading volume")

plt.show()
```

```python
df.nlargest(5, "volume")
```

```python
df["daily price change"] = df['close'] - df['open']
df["daily price change"]
```

```python
# volume vs daily price change
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)

plt.title("Trading volum vs daily price change")
plt.xlabel("Date")
plt.ylabel("Trading volume * daily price change")

plt.show()
```

```python
df["daily price range"] = df['close'] - df['open']
df["daily price range"]
```

```python
# volume vs daily price range

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price range"
)

plt.title("Trading volume vs daily price range")
plt.xlabel("Volume")
plt.ylabel("Daily price range")

plt.show()
```

```python
#Calulate the 20-Day SMA

df["SMA_20"] = df["close"].rolling(window=20).mean()
df["SMA_20"]
```

```python
#plot Closing Price + 20_Day SMA
plt.figure(figsize=(14,6))

sns.lineplot(data=df, x=df.index, y="close", label="Closing Price")
sns.lineplot(data=df, x=df.index, y="SMA_20", label="20-Day SMA")

plt.title("Shopify Closing Price and 20-Day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#Calulate the 50-Day SMA

df["SMA_50"] = df["close"].rolling(window=50).mean()
df["SMA_50"]
```

```python
df["SMA_50"] = df["close"].rolling(window=50).mean()
#plot Closing Price + 50_Day SMA
plt.figure(figsize=(14,6))

sns.lineplot(data=df, x=df.index, y="close", label="Closing Price")
sns.lineplot(data=df, x=df.index, y="SMA_50", label="50-Day SMA")

plt.title("Shopify Closing Price and 50-Day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#plot close + both moving averages

plt.figure(figsize=(14,6))
sns.lineplot(data=df, x=df.index, y="close", label="Closing Price")
sns.lineplot(data=df, x=df.index, y="SMA_20", label="20-Day SMA")
sns.lineplot(data=df, x=df.index, y="SMA_50", label="50-Day SMA")

plt.title("Shopify Closing Price, 20-Day SMA, and 50-Day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```

```python
#daily return%

df["daily returns%"] = ((df["close"] - df["open"]) / df["open"]) * 100
```

```python
df["daily returns%"].head()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="daily returns%",
    bins=20,

)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Returns")
plt.ylabel("Frequency")
plt.show()
```

```python
#volatility

df["Absolute Return"] = df["daily returns%"].abs()
df["Absolute Return"]
```

```python
df.nlargest(10,"Absolute Return")
```

```python
plt.figure(figsize=(14,5))
sns.lineplot(data=df, x="date", y="daily returns%")
plt.axhline(0,linestyle="--")
plt.xlabel("Date")
plt.ylabel("Daily Return%")
plt.title("Shopify daily returns over time")

plt.show()
```
